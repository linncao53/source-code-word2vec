```python
from github import Github
import os
import requests
import shutil
```

```python
# Get GitHub token
g = Github("MY_GITHUB_TOKEN")

repos = ["mockito/mockito", "vipshop/vjtools", "flyway/flyway", "raphw/byte-buddy", 
    "liquibase/liquibase", "rubenlagus/TelegramBots", "mongodb/mongo-java-driver", "igniterealtime/Smack", 
    "jgrapht/jgrapht", "ta4j/ta4j"]

destination_folder = "my_repos"

if not os.path.exists(destination_folder):
    os.makedirs(destination_folder)

for repo_name in repos:
    repo = g.get_repo(repo_name)
    repo_folder = os.path.join(destination_folder, repo.name)
    os.makedirs(repo_folder, exist_ok=True)
    for branch in ["main", "master"]:
        try:
            for content in repo.get_contents("", branch):
                if content.type == 'dir':
                    try:
                        os.makedirs(os.path.join(repo_folder, content.path))
                    except FileExistsError:
                        pass
                elif content.type == 'file':
                    path = content.path
                    print(f"Downloading {repo_name}/{path}")
                    try:
                        content_file = repo.get_contents(path, branch)
                        with open(os.path.join(repo_folder, path), "wb") as f:
                            f.write(content_file.content)
                    except Exception as e:
                        print(e)
            break
        except Exception as e:
            print(f"Failed to download from {branch} branch: {e}")
```

```python
java_files = []
for dirpath, dirnames, filenames in os.walk('my_repos'):
    for filename in filenames:
        if filename.endswith('.java'):
            java_files.append(os.path.join(dirpath,filename))

# print(py_files[:10])
```

```python
with open('java_files.txt', 'w', encoding='utf-8', errors='replace') as f:
    for java_file in java_files:
        try:
            with open(java_file, 'r', encoding='utf-8', errors='replace') as f_in:
                f.write(f_in.read() + '\n')
        except OSError:
            continue
```

```python
import codecs
import string
from gensim.models import Word2Vec

def tokenize_line(line):
    line = line.lower()  
    word_list = [word.strip(string.punctuation) for word in line.split()]
    word_list = list(filter(None, word_list))
    return word_list

with codecs.open('java_files.txt', 'r', 'utf-8') as f:
    lines_of_code = []
    num_lines = 0
    tokens = 0
    for line in f:
        if line.strip().startswith("#"):
            continue  
        lines_of_code.append(tokenize_line(line))
        num_lines += 1
        tokens += len(tokenize_line(line))

print("Number of lines:", num_lines)
print("Tokens:", tokens)
```

```python
model = Word2Vec(lines_of_code, window=5, min_count=1)

print("Closest words to \"for\":")
print(model.wv.most_similar("for"))

print("\nClosest words to \"if\":")
print(model.wv.most_similar("if"))

print("\nSimilarity of \"math\" and \"numpy\":")
print(model.wv.similarity("math", "numpy"))
```

```python
import ast
from collections import Counter

with codecs.open('py_files.txt', "r", "utf-8") as f:
    code = f.read()

# get a list of all identifiers in the ast tree
identifiers = []
for line in code.splitlines():
    try:
        ast_tree = ast.parse(line)
        for node in ast.walk(ast_tree):
            if isinstance(node, ast.Name):
                identifiers.append(node.id)
    except SyntaxError:
        continue
```

```python
# count the frequency
freq = Counter(identifiers)

# find the 50 most common identifiers 
with codecs.open('most_common.txt', 'w', encoding='utf-8', errors='replace') as f:
    for name, count in freq.most_common(50):
        f.write("Name: " + str(name) + ", count: " + str(count) + "\n")
```

```python
import itertools

common_ids = [name for name, count in freq.most_common(50)]

similarities = {}
for id1, id2 in itertools.combinations(common_ids, 2):
    try:
        similarity = model.wv.similarity(id1, id2)
        similarities[(id1, id2)] = similarity
    except KeyError:
        continue

print('Most similar pairs:')
sorted_similarities = sorted(similarities.items(), key=lambda x: x[1], reverse=True)
for pair, similarity in sorted_similarities[:10]:
    print(f'{pair[0]} - {pair[1]}: {similarity:.4f}')

print('\nLeast similar pairs:')
sorted_similarities = sorted(similarities.items(), key=lambda x: x[1])
for pair, similarity in sorted_similarities[:10]:
    print(f'{pair[0]} - {pair[1]}: {similarity:.4f}')
```
